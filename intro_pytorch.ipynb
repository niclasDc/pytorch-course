- Son representaciones numérica de objetos o cosas.
- Las redes neuronales toman estas representaciones para realizar las operaciones matemáticas y sacar otra representación de tensor de la probabilidad o categorización que luego son transformados a datos de entendimiento humano.


### Random Tensors 
Random tensors son importantes por la manera en que la redes neuronales aprenden, ya que inician con tensores con datos aleatorios.

```python

#* Create a random tensor of ndim 2
random_tensor = torch.rand(3,4)
random_tensor
```

```python
random_tensor.ndim
```

```python

#* Create a random tensor of ndim 3
random_tensor = torch.rand(1,3,4)
random_tensor
```

```python

#* Create a random tensor with similar shape to an image tensor
random_image_tensor = torch.rand(size=(224,224,3)) #* height, width, colour channel
random_image_tensor.shape, random_image_tensor.ndim
```

### Zeros and Ones
Ayudan para la creación de tensores que se inicialicen en cero o unos

```python

#* Create a tensor of all zeros
zeros_tensor = torch.zeros(size=(3,4))
zeros_tensor
```

```python

#* Create a tensor of all ones
ones = torch.ones(size=(3,4))
ones
```

```python
ones.dtype
```

### Creando un rango de tensores y tensors-like

```python

#* Use torch.arange para crear un tensores de rango especifico
one_to_ten = torch.arange(1, 11, 1) #* Start, stop, step
one_to_ten
```

```python

#* Para crear tensores de dimensiones iguales a otros tensores, podemos usar la función torch.zeros_like que crea un tensor de ceros con las dimensiones del tensor de entrada.
ten_zeros = torch.zeros_like(one_to_ten)
ten_zeros
```

### Tensor DataTypes

```python

#* Float 32 tensor es asignado por defecto cuando no se le pasa un dtype
float_32_tensor = torch.tensor([1.0, 2.0, 3.0], dtype=None, #* dtype es el tipo de dato del tensor
                               device=None, #* device es la GPU o CPU
                               requires_grad=False) #* requires_grad es si el tensor requiere gradientes para la optimización
print(f"Tensor: {float_32_tensor}")
print(f"Data type: {float_32_tensor.dtype}")
```

```python
float_16_tensor = torch.tensor([1.0, 2.0, 3.0], dtype=torch.float16)
print(f"Tensor: {float_16_tensor}")
print(f"Data type: {float_16_tensor.dtype}")
```

```python

#* Podemos convertir un tensor a otro tipo de dato
float_16_tensor = float_32_tensor.type(torch.float16)
print(f"Tensor: {float_16_tensor}")
print(f"Data type: {float_16_tensor.dtype}")
```

para ver que tipos de datos puede aceptar un tensor la puedes ver en esta parte [tipos de dtypes](https://pytorch.org/docs/stable/tensors.html), los distintos tipos especifican la precision computacional.
