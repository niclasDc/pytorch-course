## Que es Pytorch?

- Es uno de los más populares frameworks de investigación en deep learning.
- Permite la escritura de código rápido de aprendizaje profundo en python.
- Permita el acceso a muchos pre-built modelos de deep learning.

### Por que Pytorch?

- Es código abierto y es usado por diferentes empresas.
- Como se ve en [Code With Papers](https://paperswithcode.com/trends) es el framework más usado en papers.

## Que es un Tensor?

- Son representaciones numérica de objetos o cosas.
- Las redes neuronales toman estas representaciones para realizar las operaciones matemáticas y sacar otra representación de tensor de la probabilidad o categorización que luego son transformados a datos de entendimiento humano.
