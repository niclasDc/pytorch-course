Multiplicación de matrices

Hay dos maneras de realizar multiplicaciones en redes neuronales y deep learning:
1. Multiplicación por elemento.
2. Producto punto.

Hay dos reglas principales para realizar la multiplicación de matrices:
1. Las dimensiones internas de cada matrix deben coincidir
   * `(3, 2) @ (3, 2)` no coinciden.
   * `(2, 3) @ (3, 2)` coinciden.
   * `(3, 2) @ (2, 3)` coinciden.
2. La matrix resultado tiene forma de las dimensiones de afuera.
   * `(2, 3) @ (3, 2)` --> `(2,2)`.
   * `(3, 2) @ (2, 3)` --> `(3,3)`.

```python

#* Multiplicación elemento a elemento	
print(tensor, "*", tensor)
print(f"Equals: {tensor * tensor}")
```

```python

#* Producto Punto
torch.matmul(tensor, tensor)
```

```python
import time
```

```python

#* Tiempo demorado realizando la operación a mano
value = 0
time_start = time.time()
for i in range(len(tensor)):
    value += tensor[i] * tensor[i]
time_end = time.time()
print(value)
print(f"Time taken: {time_end - time_start} seconds")
```

```python
time_start = time.time()
value = torch.matmul(tensor, tensor)
time_end = time.time()
print(value)
print(f"Time taken: {time_end - time_start} seconds")
```

### Reshaping, stacking, squeezing y unsqueezing tensor

* Reshaping - cambia la forma o dimensiones de un tensor
* View - retorna una vista de un tensor con ciertas dimensiones pero mantiene la misma memoria como el tensor original.
* Stacking - combina varios tensores de forma horizontal (hstack)  o vertical (vstack).
* Squeezing - remueve todas `1` dimensiones de un tensor.
* Unsquezzing - añade una `1` dimension a un tensor.
* Permute - retorna una vista de un tensor con las dimensiones permutadas de cierta manera.

```python

x = torch.arange(1., 10.)
x, x.shape
```

```python

#* Añadir una dimensión extra
x_reshaped = x.reshape(9, 1)
x_reshaped, x_reshaped.shape
```

```python

#* Cambiando la vista
z = x.view(1, 9)
z, z.shape
```

```python

#* Si cambiamos algo en z también cambia en x (porque una vista comparte la misma memoria que el tensor original)
z[:, 0] = 5
z, x
```

```python

#* Stack tensor encima del otro
x_stacked = torch.stack([x, x, x, x], dim=0) #* dim=0 es el eje vertical
x_stacked, x_stacked.shape
```

```python
x_stacked = torch.stack([x, x, x, x], dim=1) #* dim=1 es el eje horizontal
x_stacked, x_stacked.shape
```

```python

#* Squeezing
x_reshaped, x_reshaped.shape
```

```python
x_squeezed = x_reshaped.squeeze() #* Squeeze elimina las dimensiones de tamaño 1 
x_squeezed, x_squeezed.shape
```

```python

#* Unsqueeze
x_squeezed
```

```python
x_unsqueezed = x_squeezed.unsqueeze(dim=0) #* Unsqueeze añade una dimensión de tamaño 1 en la posición especificada
x_unsqueezed, x_unsqueezed.shape
```

```python
x_unsqueezed = x_squeezed.unsqueeze(dim=1)
x_unsqueezed, x_unsqueezed.shape
```

```python

#* Permute
x_original = torch.rand(size=(224, 224, 3)) #* height, width, colour channel

x_permuted = x_original.permute(2, 0, 1) #* Cambia el orden de las dimensiones 0->1, 1->2, 2->0

print(f"Dimension original: {x_original.shape}")
print(f"Dimension permutada: {x_permuted.shape}") #* Colour channel, height, width
```

### Indexación en Tensores

La indexación es similar a numpy

```python
 
#* Creamos un tensor
x = torch.arange(1., 10.).reshape(1,3,3)
x, x.shape
```

```python

#* Indexación en la primera dimensión
i = 0
x[i] #* Devuelve el primer elemento del tensor en la primera dimensión
```

```python

#* Indexación en la segunda dimensión
for j in range(x.shape[1]):
    print(f"Elemento [{i},{j}] en el tensor")
    print(x[0, j]) #* Devuelve el j-ésimo elemento del tensor en la segunda dimensión
```

```python

#* Indexación en más de dos dimensiones
for j in range(x.shape[1]):
    for k in range(x.shape[2]):
        print(f"Elemento [{i},{j},{k}] en el tensor")
        print(x[0, j, k]) #* Devuelve el k-ésimo elemento del tensor en la k-ésima dimensión 

#* Se sigue la misma lógica para indexar en más dimensiones.
```

```python

#* Se puede usar : para seleccionar todos los elemetos de una dimensión
x[:, :, 0]
```

### Reproducibility

Para reducir la aleatoriedad en una red neuronal PyTorch usa el concepto de **random seed**. Esencialmente random seed lo que hace es darle un "toque" a la aleatoriedad.

```python
random_tensor_A = torch.rand(3, 4)
random_tensor_B = torch.rand(3, 4)

print(random_tensor_A)
print(random_tensor_B)
print(random_tensor_A == random_tensor_B)
```

```python

#* Vamos hacer algo random pero con tensores reproducibles
#* Configuramos un random seed
RANDOM_SEED = 23

torch.manual_seed(RANDOM_SEED) #! Funciona con un solo bloque de código
random_tensor_C = torch.rand(3, 4)

torch.manual_seed(RANDOM_SEED)
random_tensor_D = torch.rand(3, 4)

print(random_tensor_C)
print(random_tensor_D)
print(random_tensor_C == random_tensor_D) #* Ahora son iguales porque tienen la misma semilla  
```
